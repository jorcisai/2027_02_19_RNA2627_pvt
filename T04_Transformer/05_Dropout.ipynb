# 5 Dropout

[**Dropout:**](https://docs.pytorch.org/docs/stable/generated/torch.nn.Dropout.html) $\;$ capa de regularización de las salidas de otra capa
* Entrenamiento: $\;$ anula cada salida con una probabilidad dada $\,p;\;$ las salidas no anuladas se re-escalan con $\,\dfrac{1}{1-p}$
* Evaluación: $\;$ actúa como la función identidad

**Fuente:** $\;$ [Improving neural networks by preventing co-adaptation of feature detectors](https://arxiv.org/abs/1207.0580)

**Ejercicio:** $\;$ aplica dropout a $\,X=\begin{pmatrix}0.0415&2.2411\\1.2927&-0.6469\\0.9508&0.8249\end{pmatrix}\,$ en los siguientes casos
1. $p=0.1,\,$ asumiendo que no anula ninguna salida
2. $p=0.5,\,$ asumiendo que anula las salidas de la primera columna y última fila


<p style="page-break-after:always;"></p>


**Solución:**
$$\begin{pmatrix}0.0415&2.2411\\1.2927&-0.6469\\0.9508&0.8249\end{pmatrix}\frac{1}{0.9}
=\begin{pmatrix}0.0461&2.4901\\1.4364&-0.7188\\1.0564&0.9166\end{pmatrix}
\qquad\qquad%
\begin{pmatrix}0&2.2411\\0&-0.6469\\0&0\end{pmatrix}\frac{1}{0.5}
=\begin{pmatrix}0&4.4821\\0&-1.2938\\0&0\end{pmatrix}$$
