# 17 Root Mean Square Layer Normalization

[**RMSNorm (2019):**](https://arxiv.org/abs/1910.07467) $\;$ se aplica sobre las últimas dimensiones del batch

$$\operatorname{RMSNorm}(x_{ni}) = w_i\frac{x_{ni}}{\mathrm{RMS}(x_n)}, \quad \text{donde} \quad \text{RMS}(x_n) = \sqrt{\epsilon + \frac{1}{d_{\text{model}}} \sum_{i=1}^{d_{\text{model}}} x_{ni}^2}$$

**Ejemplo** ($w_i=1$): $$X=\begin{pmatrix}0.0415&2.2411\\1.2927&-0.6469\\0.9508&0.8249\end{pmatrix} \; \text{with} \; \operatorname{RMS}(x_n) = 
\begin{bmatrix}
1.5850\\%
1.0221\\%
0.8901%
\end{bmatrix}
\;\to\;%
\operatorname{RMSNorm}(X)%
=\begin{pmatrix}0.0262&1.4140\\1.2647&-0.6329\\1.0682&0.9268\end{pmatrix}$$

```python
import torch; X = torch.tensor([[0.0415, 2.2411], [1.2927, -0.6469], [0.9508, 0.8249]], dtype=float)
print(X.shape, str(X).replace('\n','').replace(' ',''))
```

```python
Y = torch.nn.RMSNorm(normalized_shape=X.shape[-1], eps=0., elementwise_affine=False, dtype=X.dtype)(X)
print(Y.shape, str(Y).replace('\n','').replace(' ',''))
```

Paso a paso:

```python
X_flat = X.flatten(start_dim=-1)
print(X_flat.shape, str(X_flat).replace('\n','').replace(' ',''))
```

```python
X_rms = X_flat.square().mean(dim=-1).reshape(-1, 1).sqrt()
print(X_rms.shape, str(X_rms).replace('\n','').replace(' ',''))
```

```python
Y_alt = (X_flat / X_rms).reshape(X.shape); Y_alt
```

```python
(Y - Y_alt).abs().max().item()
```
