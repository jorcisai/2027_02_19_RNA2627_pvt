[**Entropía cruzada:**](https://docs.pytorch.org/docs/stable/generated/torch.nn.CrossEntropyLoss.html#torch.nn.CrossEntropyLoss) $\;$ primero predicción $\,\hat{\boldsymbol{y}}\,$ y luego referencia $\,\boldsymbol{y}\,$ por similitud con PyTorch
$$\mathbb{H}(\hat{\boldsymbol{y}}, \boldsymbol{y})=-\sum_c\;y_c\log\hat{y}_c$$

**Ejemplos:** $\;$ $C=2,\,\mathcal{C}=\{0,1\}$
$$\begin{align*}
\mathbb{H}((0.2123, 0.7877)^t, (0.5738, 0.4262)^t)&=-0.5738\log 0.2123-0.4262\log 0.7877=0.9910\\%
\mathbb{H}((0.2123, 0.7877)^t, (1, 0)^t)&=-\log 0.2123=1.5498%
\end{align*}$$ 

```python
import torch; import torch.nn as nn
H = nn.CrossEntropyLoss(); torch.manual_seed(23); torch.set_printoptions(precision=4)
y_pred_logits = torch.randn(1, 2); print("y_pred_logits =", y_pred_logits)
y_pred_probs = y_pred_logits.softmax(dim=1); print("y_pred_probs =", y_pred_probs)
y_probs = torch.randn(1, 2).softmax(dim=1); print("y_probs =", y_probs)
print(f"H({y_pred_logits}, {y_probs}) = ", H(y_pred_logits, y_probs))
y_probs_argmax = y_probs.argmax(dim=1); print("y_probs_argmax =", y_probs_argmax)
print(f"H({y_pred_logits}, {y_probs_argmax}) = ", H(y_pred_logits, y_probs_argmax))
```

[**Label smoothing:**](https://arxiv.org/abs/1512.00567) $\;$ dado $\,\epsilon\in[0,1],\,$ suaviza la referencia mediante interpolación con la uniforme
$$\operatorname{LabelSmoothing}(\boldsymbol{y},\epsilon)
=(1-\epsilon)\,\boldsymbol{y}+\epsilon\,\frac{1}{C}\boldsymbol{1}_C$$

**Ejemplos (cont.):** $\;$ $\epsilon=0.1$
$$\begin{align*}
\operatorname{LabelSmoothing}((0.5738, 0.4262)^t, 0.1)&=0.9\,(0.5738, 0.4262)^t+0.1\,(0.5, 0.5)^t=(0.5664, 0.4336)^t\\
\mathbb{H}((0.2123, 0.7877)^t, (0.5664, 0.4336)^t)&=-0.5664\log 0.2123-0.4336\log 0.7877=0.9813\\%
\operatorname{LabelSmoothing}((1, 0)^t, 0.1)&=0.9\,(1, 0)^t+0.1\,(0.5, 0.5)^t=(0.95, 0.05)^t\\
\mathbb{H}((0.2123, 0.7877)^t, (0.95, 0.05)^t)&=-0.95\log 0.2123-0.05\log 0.7877=1.4842%
\end{align*}$$ 

```python
import torch; import torch.nn as nn
H = nn.CrossEntropyLoss(label_smoothing=0.1); torch.manual_seed(23); torch.set_printoptions(precision=4)
y_pred_logits = torch.randn(1, 2); print("y_pred_logits =", y_pred_logits)
y_pred_probs = y_pred_logits.softmax(dim=1); print("y_pred_probs =", y_pred_probs)
y_probs = torch.randn(1, 2).softmax(dim=1); print("y_probs =", y_probs)
print(f"H({y_pred_logits}, {y_probs}) = ", H(y_pred_logits, y_probs))
y_probs_argmax = y_probs.argmax(dim=1); print("y_probs_argmax =", y_probs_argmax)
print(f"H({y_pred_logits}, {y_probs_argmax}) = ", H(y_pred_logits, y_probs_argmax))
```
