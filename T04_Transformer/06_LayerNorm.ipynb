# 6 LayerNorm

[**LayerNorm:**](https://docs.pytorch.org/docs/stable/generated/torch.nn.LayerNorm.html) $\;$ capa que estandariza cada dato en su última dimensión. Por ejemplo, para un tensor bidimensional:

$$
\operatorname{LayerNorm}(x_{ni}) = w_i \frac{x_{ni} - \hat{\mu}_n}{\hat{\sigma}_n} + b_i 
$$

donde $\hat{\mu}_n$ y $\hat{\sigma}_n$ son la media y desviación típica, respectivamente, calculadas en su última dimensión.

**Ejemplo** ($w_i=1$ y $b_i=0$): $$X=\begin{pmatrix}0.0415&2.2411\\1.2927&-0.6469\\0.9508&0.8249\end{pmatrix} \; \text{with} \;
\begin{bmatrix}
\hat{\mu}_1=1.1413&\hat{\sigma}_1=1.5553\\%
\hat{\mu}_2=0.3229&\hat{\sigma}_2=1.3715\\%
\hat{\mu}_3=0.8879&\hat{\sigma}_3=0.0890%
\end{bmatrix}
\;\to\;%
\operatorname{LayerNorm}(X)%
=\begin{pmatrix}-0.7071&0.7071\\0.7071&-0.7071\\0.7070&-0.7070\end{pmatrix}$$
