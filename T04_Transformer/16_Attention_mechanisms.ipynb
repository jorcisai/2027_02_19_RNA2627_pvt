# 16 Mecanismos de atención

**[Atención multiconsulta (2019)](https://arxiv.org/pdf/1911.02150):**
$\;W_i^Q\in\mathbb{R}^{d_{\text{model}}\times d_{\text{k}}},\,$
$\;W^K\in\mathbb{R}^{d_{\text{model}}\times d_{\text{k}}},\,$
$W^V\in\mathbb{R}^{d_{\text{model}}\times d_{\text{v}}},\,$
$W^O\in\mathbb{R}^{h d_{\text{v}}\times d_{\text{model}}}$
$$\operatorname{MultiQuery}(Q,K,V)=\operatorname{Concat}(\operatorname{head}_1,\dotsc,\operatorname{head}_h)W^O\in\mathbb{R}^{n\times d_{\text{model}}}$$
$$\operatorname{head}_i=\operatorname{Attention}(QW_i^Q,KW^K,VW^V)\in\mathbb{R}^{n\times d_{\text{v}}}$$

**[Atención consulta agrupada (2023)](https://arxiv.org/pdf/2305.13245):** Generalización de atención multiconsulta

Disponemos de $g$ grupos de $K$ y $V$, y asignamos cada consulta $i$ a un grupo $j$ de par clave-valor mediante una función $\gamma(i)$.

$$\;W_i^Q\in\mathbb{R}^{d_{\text{model}}\times d_{\text{k}}},\,
\;W_j^K\in\mathbb{R}^{d_{\text{model}}\times d_{\text{k}}},\,
W_j^V\in\mathbb{R}^{d_{\text{model}}\times d_{\text{v}}},\,
W^O\in\mathbb{R}^{h d_{\text{v}}\times d_{\text{model}}}$$
$$\operatorname{GroupedQuery}(Q,K,V)=\operatorname{Concat}(\operatorname{head}_1,\dotsc,\operatorname{head}_h)W^O\in\mathbb{R}^{n\times d_{\text{model}}}$$
$$\operatorname{head}_i=\operatorname{Attention}(QW_i^Q,KW_{\gamma(i)}^K,VW_{\gamma(i)}^V)\in\mathbb{R}^{n\times d_{\text{v}}}$$

Casos de interés: atención multiconsulta ($g=1$) y atención multicabeza ($g=h$)

**[Atención con puertas (2025)](https://arxiv.org/pdf/2505.06708):** 
El uso de puertas se formaliza como:
$$
X' = g(X,Y,W^G,\sigma) = X \odot \sigma(YW^G)
$$
donde
- $X$ es la entrada que va a ser modulada
- $Y$ es la entrada usada para calcular los valores de la puerta (i.e. estados ocultos normalizados)
- $W^G$ son los parámetros de la puerta
- $\sigma$ es una función de activación (i.e., sigmoid)
- $\odot$ es una operación elemento a elemento (i.e. producto Hadamard)
