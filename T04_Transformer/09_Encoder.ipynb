# 11 Encoder

**Auto-atención:** $\quad$ $\operatorname{SelfAttention}(X)=\operatorname{MultiHead}(X,X,X)$

**Capa encoder:** $\;$ dada una secuencia de entrada $\,X=X^{(1)}\in\mathbb{R}^{n\times d_{\text{model}}},\,$ aplica
$$\begin{align*}
X^{(2)}&=\operatorname{LayerNorm}(X^{(1)})\\
X^{(3)}&=\operatorname{SelfAttention}(X^{(2)})\\
X^{(4)}&=\operatorname{Dropout}(X^{(3)})\\
X^{(5)}&=\operatorname{Add}(X^{(1)},X^{(4)})\\
X^{(6)}&=\operatorname{LayerNorm}(X^{(5)})\\
X^{(7)}&=\operatorname{PWFFN}(X^{(6)})\\
X^{(8)}&=\operatorname{Dropout}(X^{(7)})\\
\operatorname{EncoderLayer}(X)=X^{(9)}&=\operatorname{Add}(X^{(5)},X^{(8)})
\end{align*}$$


**Pre-norm vs post-norm:** $\;$ LayerNorm primero (pre-norm; arriba) o tras Add (post-norm; versión original)

**Encoder:** $\;$ dada una secuencia de entrada $\,X=X^{(1)}\in\mathbb{R}^{n\times d_{\text{model}}},\,$ aplica una pila de $N=6$ capas encoder idénticas
$$\begin{align*}
X^{(2)}&=\operatorname{EncoderLayer}(X^{(1)})\\
X^{(3)}&=\operatorname{EncoderLayer}(X^{(2)})\\
\vdots\\
X^{(N+1)}&=\operatorname{EncoderLayer}(X^{(N)})\\
\operatorname{Encoder}(X)&=\operatorname{LayerNorm}(X^{(N+1)})
\end{align*}$$

**Ejemplo:** $\;$ $\operatorname{Encoder}(X)\,$ de una capa con la entrada y parámetros vistos anteriormente
$$\begin{align*}

X^{(6)}&=\operatorname{LayerNorm}(X^{(5)})=\begin{pmatrix}-0.7071&0.7071\\0.7071&-0.7071\\0.7070&-0.7070\end{pmatrix}\\
X^{(7)}&=\operatorname{PWFFN}(X^{(6)})=\begin{pmatrix}0.2896&-0.1471\\1.0716&0.3682\\0.2896&-0.1470\end{pmatrix}\\

\operatorname{Encoder}(X)&=\operatorname{LayerNorm}(X^{(9)})=\begin{pmatrix}-0.7071&0.7071\\0.7071&-0.7071\\0.7071&-0.7071\end{pmatrix}
\end{align*}$$
