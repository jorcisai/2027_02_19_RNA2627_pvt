**Example group norm:** $\;$ input $\;\mathbf{X}\,N\times C\times x_h\times x_w\;$ and output $\;\mathbf{Y}\,N\times C\times x_h\times x_w$

$$\mathbf{X}=\begin{bmatrix}
  \begin{bmatrix} % N 1
    \begin{bmatrix} % C 1
        0
    \end{bmatrix},
    \begin{bmatrix} % C 2
        1
    \end{bmatrix},
    \begin{bmatrix} % C 3
        2
    \end{bmatrix},
    \begin{bmatrix} % C 4
        3
    \end{bmatrix}
  \end{bmatrix}
\end{bmatrix}$$

$$\mathbf{Y}=\operatorname{GroupNorm}(\mathbf{X}, 2)=\begin{bmatrix}
  \begin{bmatrix} % N 1
    \begin{bmatrix} % C 1
        -1
    \end{bmatrix},
    \begin{bmatrix} % C 2
        1
    \end{bmatrix},
    \begin{bmatrix} % C 3
        -1
    \end{bmatrix},
    \begin{bmatrix} % C 4
        1
    \end{bmatrix}
  \end{bmatrix}
\end{bmatrix}$$

```python
import torch; X = torch.tensor([ [ [[0]], [[1]], [[2]], [[3]] ] ], dtype=float)
print(X.shape, str(X).replace('\n','').replace(' ',''))
```

```python
Y = torch.nn.GroupNorm(num_groups=2, num_channels=4, eps=0., affine=False, dtype=X.dtype)(X)
print(Y.shape, str(Y).replace('\n','').replace(' ',''))
```

Step by step: <!-- https://github.com/pytorch/pytorch/blob/main/torch/_refs/__init__.py#L3339 -->

```python
X_flat = X.reshape(X.shape[0], 2, X.shape[1]//2, *X.shape[2:]).flatten(start_dim=2)
print(X_flat.shape, str(X_flat).replace('\n','').replace(' ',''))
```

```python
X_mean = X_flat.mean(dim=2).reshape(-1, 1)
print(X_mean.shape, str(X_mean).replace('\n','').replace(' ',''))
```

```python
X_std = X_flat.std(dim=2, correction=0).reshape(-1, 1)
print(X_std.shape, str(X_std).replace('\n','').replace(' ',''))
```

```python
Y_alt = ((X_flat - X_mean) / X_std).reshape(X.shape); (Y - Y_alt).abs().max().item()
```
