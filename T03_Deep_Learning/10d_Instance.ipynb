**Example instance norm:** $\;$ input $\;\mathbf{X}\,N\times C\times x_h\times x_w\;$ and output $\;\mathbf{Y}\,N\times C\times x_h\times x_w$

$$\mathbf{X}=\begin{bmatrix}
  \begin{bmatrix} % N 1
    \begin{bmatrix} % C 1
        0&2\\2&0
    \end{bmatrix}
  \end{bmatrix}
\end{bmatrix}$$

$$\mathbf{Y}=\operatorname{InstanceNorm2d}(\mathbf{X})
=\begin{bmatrix}
  \begin{bmatrix} % N 1
    \begin{bmatrix} % C 1
        -1&1\\1&-1
    \end{bmatrix}
  \end{bmatrix}
\end{bmatrix}$$

```python
import torch; X = torch.tensor([ [ [ [0, 2], [2, 0] ] ] ], dtype=float)
print(X.shape, str(X).replace('\n',''))
```

```python
Y = torch.nn.InstanceNorm2d(num_features=1, eps=0., momentum=0., affine=False, dtype=X.dtype)(X)
print(Y.shape, str(Y).replace('\n',''))
```

Step by step:

```python
X_flat = X.flatten(start_dim=2)
print(X_flat.shape, str(X_flat).replace('\n','').replace(' ',''))
```

```python
X_mean = X_flat.mean(dim=2).reshape(-1, 1)
print(X_mean.shape, str(X_mean).replace('\n','').replace(' ',''))
```

```python
X_std = X_flat.std(dim=2, correction=0).reshape(-1, 1)
print(X_std.shape, str(X_std).replace('\n','').replace(' ',''))
```

```python
Y_alt = ((X_flat - X_mean) / X_std).reshape(X.shape); (Y - Y_alt).abs().max().item()
```
