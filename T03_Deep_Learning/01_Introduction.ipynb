# 1 Introduction

**Deep Learning:** $\quad$ Area of ​​ML specializing in learning networks with (many) hidden layers

**Deep Learning Revolution:** $\quad$ Deep learning gained widespread popularity starting in the 2010s, primarily due to improved results in automatic speech recognition (2011) and image classification (2012)

**Objective:** $\quad$ To introduce prominent deep learning architectures, primarily for image classification

**LeNet (1998):** $\quad$ Pioneering convolutional neural network (CNN), introduced in 1998 by Yann LeCun for MNIST
* Proposes the use of convolutional layers and pooling

**AlexNet (2012):** $\quad$ Alex Krizhevsky's CNN for ImageNet, which attracted considerable interest
* Spectacular results thanks to its depth, use of GPUs, ReLUs, and dropout

**Inception (2014):** Google's CNN that introduces the typical meta-architecture of modern CNNs

**ResNet (2015):** Microsoft's CNN that introduces residual connections for training very deep models
* Proposes the use of residual connections and adds normalization layers

**Transformer (2017):** Google's architecture for machine translation based on the attention mechanism
* Considered the foundation of today's most competitive AI models

**Vision Transformer (2020):** An adaptation of Transformer for vision
