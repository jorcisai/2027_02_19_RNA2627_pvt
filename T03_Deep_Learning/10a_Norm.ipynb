# 10 Layer normalization

**Objective:** $\;$ add extra layers to avoid the problem of gradients disappearing or exploding in deep networks

**Batch/layer/instance/group norm:** $\;$ $\gamma=1\,$ and $\,\beta=0\,$ can be learned; $\,\epsilon=10^{-5}\,$ by default

$$y = \frac{x - \mathrm{E}[x]}{ \sqrt{\mathrm{Var}[x] + \epsilon}} \, \gamma + \beta$$
