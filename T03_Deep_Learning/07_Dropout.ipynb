# 7 Dropout

[**Dropout:**](https://en.wikipedia.org/wiki/Dilution_\(neural_networks\)) $\;$ regularization layer of the outputs of another layer (inputs for dropout)
* Training: $\;$ cancels each input with a given probability $\,p;\;$ non-canceled inputs are rescaled with $\,\dfrac{1}{1-p}$
* Evaluation: $\;$ acts as the identity function

**Example:** $\;$ dropout applied to $\,\mathbf{X}=\begin{pmatrix}1\\4\end{pmatrix}\,$ in the following cases

1. $p=0.1,\,$ assuming it does not cancel any entry:
$\displaystyle\quad\operatorname{Dropout}(\mathbf{X}, 0.1)=\begin{pmatrix}1\\4\end{pmatrix}\frac{1}{0.9}=\begin{pmatrix}1.1111\\4.4444\end{pmatrix}$

2. $p=0.5,\,$ assuming it cancels the first entry:
$\displaystyle\quad\operatorname{Dropout}(\mathbf{X}, 0.5)=\begin{pmatrix}0\\4\end{pmatrix}\frac{1}{0.5}=\begin{pmatrix}0\\8\end{pmatrix}$

```python
import torch; torch.random.manual_seed(23); X = torch.tensor([1, 4], dtype=torch.float)
print(torch.nn.Dropout(p=0.1)(X), torch.nn.Dropout(p=0.5)(X))
```
