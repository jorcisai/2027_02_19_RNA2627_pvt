[**RMSNorm:**](https://docs.pytorch.org/docs/stable/generated/torch.nn.RMSNorm.html) $\;$ [Root Mean Square Layer Normalization](https://arxiv.org/abs/1910.07467) is applied to the last dimensions of the batch

$$y_i = \frac{x_i}{\mathrm{RMS}(x)} * \gamma_i, \quad \text{donde} \quad \text{RMS}(x) = \sqrt{\epsilon + \frac{1}{n} \sum_{i=1}^{n} x_i^2}$$

**Example:** $\;$ input $\;\mathbf{X}\,N\times C\times x_h\times x_w\;$ and output $\;\mathbf{Y}\,N\times C\times x_h\times x_w\,$ to the last dimension

$$\mathbf{X}=%
\begin{bmatrix} % N
  \begin{bmatrix} % C
    \begin{bmatrix}
        1&1\\
        2&2
    \end{bmatrix}
  \end{bmatrix}
\end{bmatrix}
\qquad%
\mathbf{Y}=\operatorname{RMSNorm}(\mathbf{X}, (2,))=%
\begin{bmatrix}
  \begin{bmatrix}
    \begin{bmatrix}
        1&1\\
        1&1
    \end{bmatrix}
  \end{bmatrix}
\end{bmatrix}$$


```python
import torch; X = torch.tensor([ [ [ [1, 1], [2, 2] ] ] ], dtype=float)
print(X.shape, str(X).replace('\n','').replace(' ',''))
```

```python
Y = torch.nn.RMSNorm(normalized_shape=X.shape[-1:], eps=0., elementwise_affine=False, dtype=X.dtype)(X)
print(Y.shape, str(Y).replace('\n','').replace(' ',''))
```

Step by step:

```python
X_flat = X.flatten(start_dim=-1)
print(X_flat.shape, str(X_flat).replace('\n','').replace(' ',''))
```

```python
X_rms = X_flat.square().mean(dim=-1).reshape(-1, 1).sqrt()
print(X_rms.shape, str(X_rms).replace('\n','').replace(' ',''))
```

```python
Y_alt = (X_flat / X_rms).reshape(X.shape); (Y - Y_alt).abs().max().item()
```
